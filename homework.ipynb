```python
import numpy as np
X = (
df[df['origin'] == 'Asia']
[['vehicle_weight', 'model_year']]
.head(7)
.values
)
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
w.sum()
```

```python
print(f"sum of weights: {w.sum()}")
```
